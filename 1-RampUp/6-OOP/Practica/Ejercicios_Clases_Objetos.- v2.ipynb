## Ejercicio 1
Somos una empresa que tiene varias tiendas de electrodomésticos. Necesitamos un programa para manejar las tiendas, ver las ventas que han tenido, cuántos empleados hay, etc... Para ello vamos a modelizar la tienda en una clase, que tendrá las siguientes características:
* Nombre clase: "Tienda"
* Atributos comunes:
    * Tipo: "Electrodomésticos"
    * Abierta: True
* Atributos propios de cada tienda:
    * Nombre - String
    * Dirección - String
    * Número de empleados - int
    * Ventas últimos 3 meses - Lista de 3 numeros
* Método para calcular las ventas de todos los meses, que devuelva un número con todas las ventas.
* Método que calcula la media de ventas de los últimos meses, por empleado
* Método que devuelve en un string el nombre de la tienda, junto con su dirección.
* Método que obtiene las ventas del último mes.
* Método para dar una proyección de las ventas en caso de haber invertido X dinero en marketing en esa tienda. Siendo X el parámetro de entrada. Si X es menor de 1000, las ventas de los ultimos 3 meses hubiesen sido de (1.2 x ventas), si es mayor o igual de 1000, las ventas hubiesen sido de (1.5 x venta). El método reescribe el atributo de ventas en los últimos 3 meses, y además devuelve las ventas con el aumento.

Se pide:
1. Implementar la clase "Tienda"
2. Crear tres tiendas con datos inventados
3. Comprueba en al menos una de ellas, todo lo implementado en la clase tienda (sus atributos, media de ventas, ventas/empleado...)
3. Calcular las ventas del último mes de todas las tiendas. Para ello usa el bucle `for`
4. Imprime por pantalla los nombres de las tiendas cuya dirección lleve el string "Avenida"
5. Seguro que ya has hecho este apartado... Documenta la clase :)

1, 2 y 5. Implementación de la clase Tienda (Documentada) y Creación de Objetos

```python
class Tienda:
    """
    Clase que representa una tienda de electrodomésticos para gestionar sus ventas,
    empleados, datos de contacto y proyecciones de marketing.

    Atributos de clase:
    -------------------
    tipo : str
        El sector de la tienda, por defecto "Electrodomésticos".
    abierta : bool
        Indica si la tienda está operativa, por defecto True.
    """
    tipo = "Electrodomésticos"
    abierta = True

    def __init__(self, nombre, direccion, num_empleados, ventas):
        """
        Constructor de la clase Tienda.

        Parámetros:
        -----------
        nombre : str
            Nombre identificativo de la sucursal.
        direccion : str
            Dirección física de la tienda.
        num_empleados : int
            Cantidad de trabajadores contratados.
        ventas : dict
            Diccionario anidado que contiene el histórico de ventas.
            Estructura: { "Nombre Empleado": { "Mes": importe_entero } }
        """
        self.nombre = nombre
        self.direccion = direccion
        self.num_empleados = num_empleados
        self.ventas = ventas

    def calcular_ventas_totales(self):
        """Calcula y devuelve el importe sumado de todos los meses de todos los empleados."""
        importe_total = 0 
        for empleado, meses in self.ventas.items():
            importe_total += sum(meses.values())
        return importe_total

    def calcular_media_ventas_empleado(self):
        """Calcula la media de ventas de los últimos 3 meses para cada uno de los empleados."""
        medias = {}
        for empleado, meses in self.ventas.items():
            lista_ventas = list(meses.values())
            ventas_ultimos_meses = lista_ventas[-3:]
    
            total_ventas = sum(ventas_ultimos_meses)
            cantidad_meses = len(ventas_ultimos_meses)
            
            if cantidad_meses > 0:
                medias[empleado] = total_ventas / cantidad_meses
            else:
                medias[empleado] = 0
        return medias
    
    def obtener_nombre_direccion_tienda(self):
         """Devuelve una cadena formateada con el nombre y la localización de la sucursal."""
         return f"Nombre: {self.nombre}\nDirección: {self.direccion}"

    def obtener_ventas_ultimo_mes(self):
        """Suma y devuelve los ingresos generados únicamente en el último mes registrado por todos los empleados."""
        total_ventas_tienda = 0
        for empleado, meses in self.ventas.items():
             lista_ventas = list(meses.values())
             if lista_ventas:
                 ventas_ultimo_mes = lista_ventas[-1]
                 total_ventas_tienda += ventas_ultimo_mes
        return total_ventas_tienda
    
    def proyectar_y_actualizar_ventas(self, x_marketing):
        """
        Simula una campaña de marketing sobre los últimos 3 meses basándose en el presupuesto X.
        Si X < 1000 incrementa un 20% (x1.2). Si X >= 1000 incrementa un 50% (x1.5).
        Modifica el atributo de ventas de forma permanente y devuelve la simulación.
        """
        if x_marketing < 1000:
            multiplicador = 1.2
        else:
            multiplicador = 1.5
        
        proyeccion_final = {}
        for empleado, meses in self.ventas.items():
            lista_meses = list(meses.keys())
            ultimos_tres_meses = lista_meses[-3:]
            proyeccion_final[empleado] = {}
            
            for mes in ultimos_tres_meses:
                venta_proyectada = meses[mes] * multiplicador
                meses[mes] = venta_proyectada
                proyeccion_final[empleado][mes] = venta_proyectada
        return proyeccion_final


# ==========================================
# 2. CREACIÓN DE TRES TIENDAS CON DATOS
# ==========================================

ventas_t1 = {
    "Ana Gómez": {"Octubre": 1000, "Noviembre": 2000, "Diciembre": 3000},
    "Luis Martínez": {"Octubre": 1500, "Noviembre": 2500, "Diciembre": 3500}
}

ventas_t2 = {
    "Roberto Díaz": {"Octubre": 2000, "Noviembre": 2000, "Diciembre": 4000},
    "Lucía Castro": {"Octubre": 3000, "Noviembre": 1000, "Diciembre": 5000}
}

ventas_t3 = {
    "Diego Flores": {"Octubre": 1200, "Noviembre": 1800, "Diciembre": 2500}
}

# Instancias
tienda_principal = Tienda("Electro Tienda Principal", "Avenida de la Constitución 45", 2, ventas_t1)
tienda_norte = Tienda("Electro Tienda Norte", "Calle de los Olmos 12", 2, ventas_t2)
tienda_sur = Tienda("Electro Tienda Sur", "Avenida Marítima 89", 1, ventas_t3)

# Agrupamos las tiendas en una lista para facilitar los bucles solicitados
todas_las_tiendas = [tienda_principal, tienda_norte, tienda_sur]
```

3. Comprobación de todas las propiedades en una tienda (tienda_principal)

```python
print("=== COMPROBACIÓN DE PROPIEDADES EN TIENDA PRINCIPAL ===")
# Atributos comunes y específicos
print(f"Tipo de negocio: {tienda_principal.tipo}")
print(f"¿Está abierta?: {tienda_principal.abierta}")
print(f"Empleados activos: {tienda_principal.num_empleados}")

# Nombre y dirección
print("\n[Método obtener_nombre_direccion_tienda]:")
print(tienda_principal.obtener_nombre_direccion_tienda())

# Ventas totales
print(f"\n[Método calcular_ventas_totales]: {tienda_principal.calcular_ventas_totales()}€")

# Media de ventas por empleado (últimos 3 meses)
print("\n[Método calcular_media_ventas_empleado]:")
medias = tienda_principal.calcular_media_ventas_empleado()
for emp, med in medias.items():
    print(f"  - {emp}: {med:.2f}€")

# Ventas del último mes
print(f"\n[Método obtener_ventas_ultimo_mes]: {tienda_principal.obtener_ventas_ultimo_mes()}€")

# Proyección de Marketing (Inversión: 1500€ -> aplica x1.5)
print("\n[Método proyectar_y_actualizar_ventas] - Inversión 1500€:")
proyeccion = tienda_principal.proyectar_y_actualizar_ventas(1500)
print(f"  - Datos de proyección devueltos: {proyeccion}")
print(f"  - Atributo modificado en la clase: {tienda_principal.ventas['Ana Gómez']}")
print("\n" + "="*50 + "\n")
```

4. Calcular las ventas del último mes de todas las tiendas usando un bucle for

```python
print("=== VENTAS DEL ÚLTIMO MES EN TODAS LAS TIENDAS ===")
for tienda in todas_las_tiendas:
    ventas_mes = tienda.obtener_ventas_ultimo_mes()
    print(f"La sucursal '{tienda.nombre}' vendió {ventas_mes}€ en el último mes.")
print("\n" + "="*50 + "\n")
```

5. Imprimir tiendas cuya dirección contenga el string "Avenida"

```python
print("=== FILTRADO DE TIENDAS POR DIRECCIÓN ('Avenida') ===")
for tienda in todas_las_tiendas:
    # Usamos .lower() para evitar problemas si está escrito en mayúsculas o minúsculas
    if "avenida" in tienda.direccion.lower():
        print(f"📍 Coincidencia encontrada: {tienda.nombre} ({tienda.direccion})")
```

## Ejercicio 2
En este ejercicio vamos a implementar una clase *Perro* en Python. La clase tiene las siguientes características:
* Cosas que sabemos seguro que tiene un perro
    * Tiene 4 patas
    * 2 orejas
    * 2 ojos
    * Una velocidad de 0. Por defecto, el perro está parado
* Cuando se inicialice:
    * El perro será de una determinada raza
    * Por defecto tendrá pelo "Marrón", a no ser que se diga lo contrario.
    * Por defecto no tendrá dueño, a no ser que se diga lo contrario.
    
* Dispondrá también de un método llamado andar, que tiene un argumento de entrada (aumento_velocidad). Este valor se le sumará a la velocidad que ya llevaba el perro.
* Necesita otro método (parar), donde pondremos la velocidad a 0.
* Otro método llamado "ladrar", que tendrá un argumento de entrada, y la salida será el siguiente string: "GUAU!" + el argumento de entrada.


Se pide:
* Implementa la clase *Perro*
* Crea un objeto de tipo *Perro*, sin dueño
* Comprueba que están bien todos sus atributos
* Prueba que ande, y comprueba su velocidad
* Páralo
* Documenta la clase *Perro*

```python
class Perro:
    """
    Clase que representa a un perro, sus características físicas básicas
    y sus acciones principales como andar, parar o ladrar.

    Atributos de clase (comunes para todos los perros):
    --------------------------------------------------
    patas : int
        Número de patas del perro (Fijo: 4).
    orejas : int
        Número de orejas del perro (Fijo: 2).
    ojos : int
        Número de ojos del perro (Fijo: 2).

    Atributos de instancia (propios de cada perro):
    -----------------------------------------------
    raza : str
        La raza específica del perro.
    pelo : str
        El color del pelo del perro (Por defecto: "Marrón").
    dueno : str | None
        El nombre del dueño del perro (Por defecto: None / Sin dueño).
    velocidad : int | float
        La velocidad actual del perro (Inicialmente: 0).
    """

    patas = 4
    orejas = 2
    ojos = 2

    def __init__(self, raza, pelo="Marrón", dueno=None):
        """Inicializa un nuevo perro con su raza, pelo y dueño."""
        self.raza = raza
        self.pelo = pelo
        self.dueno = dueno
        self.velocidad = 0  # Por defecto, el perro está parado

    def andar(self, aumento_velocidad):
        """Suma el valor recibido a la velocidad actual del perro."""
        self.velocidad += aumento_velocidad

    def parar(self):
        """Detiene al perro restableciendo su velocidad a 0."""
        self.velocidad = 0

    def ladrar(self, argumento_entrada):
        """Devuelve un string combinando un ladrido con el texto recibido."""
        return "GUAU! " + str(argumento_entrada)


# ======================================================
# 1. Crear un objeto de tipo Perro, sin dueño
# ======================================================
mi_perro = Perro("Pastor Alemán")


# ======================================================
# 2. Comprobar que están bien todos sus atributos
# ======================================================
print("=== COMPROBACIÓN DE ATRIBUTOS DEL PERRO ===")
print(f"Patas: {mi_perro.patas}")
print(f"Orejas: {mi_perro.orejas}")
print(f"Ojos: {mi_perro.ojos}")
print(f"Raza: {mi_perro.raza}")
print(f"Color de pelo: {mi_perro.pelo}")
print(f"Dueño: {mi_perro.dueno} (Indica que no tiene)")
print(f"Velocidad inicial: {mi_perro.velocidad}\n")


# ======================================================
# 3. Prueba que ande y comprueba su velocidad
# ======================================================
print("=== PROBANDO EL MÉTODO ANDAR ===")
mi_perro.andar(5)
print(f"El perro empieza a caminar. Velocidad actual: {mi_perro.velocidad}")
mi_perro.andar(3.5)
print(f"El perro acelera. Nueva velocidad actual: {mi_perro.velocidad}\n")


# ======================================================
# 4. Páralo
# ======================================================
print("=== PROBANDO EL MÉTODO PARAR ===")
mi_perro.parar()
print(f"El perro se ha detenido. Velocidad actual: {mi_perro.velocidad}\n")


# Extra: Probar el método ladrar
print("=== PROBANDO EL MÉTODO LADRAR ===")
print(mi_perro.ladrar("Tengo hambre!"))
```
