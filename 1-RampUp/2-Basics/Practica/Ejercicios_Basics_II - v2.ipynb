## Ejercicio 1
* Crea dos variables numericas: un `int` y un `float`
* Comprueba sus tipos
* Sumalas en otra nueva
* ¿De qué tipo es la nueva variable?
* Elimina las variables creadas

````python
# 1. Creamos las dos variables numéricas
variable_entera = 10         # Tipo int
variable_decimal = 5.75      # Tipo float

# 2. Comprobamos sus tipos imprimiéndolos en pantalla
print("--- Tipos de las variables originales ---")
print(f"'variable_entera' es de tipo: {type(variable_entera)}")
print(f"'variable_decimal' es de tipo: {type(variable_decimal)}")

# 3. Las sumamos en una nueva variable
resultado_suma = variable_entera + variable_decimal

# 4. Comprobamos el tipo de la nueva variable
print("\n--- Resultado y su tipo ---")
print(f"El resultado de la suma es: {resultado_suma}")
print(f"La variable 'resultado_suma' es de tipo: {type(resultado_suma)}")

# 5. Eliminamos las variables creadas de la memoria del sistema
del variable_entera
del variable_decimal
del resultado_suma
````

## Ejercicio 2
Escribe un programa para pasar de grados a radianes. Hay que usar `input`. Recuerda que la conversión se realiza mediante

$$radianes = grados*(pi/180)$$

````python
import math
````

````python
# 1. Solicitamos los grados al usuario y los convertimos a decimal (float)
grados = float(input("Introduce los grados a convertir: "))

# 2. Aplicamos la fórmula matemática usando math.pi
radianes = grados * (math.pi / 180)

# 3. Mostramos el resultado en pantalla redondeado a 4 decimales
print("\n📐 Resultado de la conversión:")
print(f"{grados}° equivalen a {radianes:.4f} radianes")
````

## Ejercicio 3
Escribe un programa que calcule el area de un paralelogramo (base x altura). También con `input`

````python
# 1. Solicitamos la base y la altura al usuario convirtiéndolas a decimales (float)
base = float(input("Introduce la longitud de la base (en cm o metros): "))
altura = float(input("Introduce la longitud de la altura (en cm o metros): "))

# 2. Aplicamos la fórmula matemática: Área = Base × Altura
area = base * altura

# 3. Mostramos el resultado por pantalla con un máximo de 2 decimales
print("\n📐 Resultado del cálculo:")
print(f"El área del paralelogramo con base {base} y altura {altura} es de {area:.2f} unidades cuadradas.")
````

## Ejercicio 8
Razona sin ejecutar código el output que obtendremos de las siguientes sentencias
1. True and True and False
2. not ((True or False) and (True or False))
3. (False or False or False or True) and False
4. not False and True and not True

Compruébalo después con python

1. False
2. not (True) and (True) = False and True = False
3. True and False = False
4. True and True and False = False

### 🧠 Razonamiento Lógico de Expresiones Booleanas

1. **`True and True and False`**
   * **Output:** `False`
   * **Razón:** El operador `and` evalúa en cadena y requiere que **todas** las condiciones sean verdaderas. Al llegar al último `and False`, la expresión se invalida por completo de izquierda a derecha.

2. **`not ((True or False) and (True or False))`**
   * **Output:** `False`
   * **Razón:** Evaluando los paréntesis internos, `(True or False)` es `True` en ambos lados. La expresión se reduce a `not (True and True)`. Como el interior del paréntesis es `True`, la negación exterior `not True` devuelve `False`.

3. **`(False or False or False or True) and False`**
   * **Output:** `False`
   * **Razón:** El primer bloque entre paréntesis se evalúa como `True` porque contiene un único elemento verdadero en una cadena de `or`. Sin embargo, la expresión final queda como `True and False`, lo que por cortocircuito devuelve `False`.

4. **`not False and True and not True`**
   * **Output:** `False`
   * **Razón:** Por precedencia de operadores, se resuelven primero las negaciones: `not False` pasa a ser `True` y `not True` pasa a ser `False`. La expresión se transforma en `True and True and False`, resultando en `False`.

````python
# Ejecución directa de las sentencias booleanas
print(f"1. Resultado: {True and True and False}")
print(f"2. Resultado: {not ((True or False) and (True or False))}")
print(f"3. Resultado: {(False or False or False or True) and False}")
print(f"4. Resultado: {not False and True and not True}")
````

## Ejercicio 10
Para el siguiente string se pide imprimir por pantalla los siguientes casos

    "Lo primero de todo, cómo están los máquinas"

1. Pasarlo todo a mayúsculas
2. Pasarlo todo a minúsculas
3. Pasar a mayúsculas solo las iniciales de las palabras
4. Crea una lista dividiendolo por sus espacios
5. Sustituye las comas `,` por puntos y comas `;`
6. Elimina las `a` minúsculas

````python
texto = "Lo primero de todo, cómo están los máquinas"

# 1. Pasarlo todo a mayúsculas
print(texto.upper())

# 2. Pasarlo todo a minúsculas
print(texto.lower())

# 3. Pasar a mayúsculas solo las iniciales de las palabras
print(texto.title())

# 4. Crea una lista dividiéndolo por sus espacios
print(texto.split(" "))

# 5. Sustituye las comas por puntos y comas
print(texto.replace(",", ";"))

# 6. Elimina las 'a' minúsculas (sin afectar a la 'á' con tilde)
print(texto.replace("a", ""))
````

## Ejercicio 11
1. Crea una lista con 3 elementos numéricos
2. Añade un cuarto elemento
3. Calcula la suma de todos
4. Elimina el segundo elemento de la lista
5. Añade otro elemento en la posicion 3 de la lista
6. Crea otra lista con 4 elementos y concaténala a la que ya tenías.

````python
# 1. Crea una lista con 3 elementos numéricos
lista = [10, 20, 30]

# 2. Añade un cuarto elemento al final
lista.append(40)

# 3. Calcula la suma de todos los elementos actuales (10 + 20 + 30 + 40)
suma_total = sum(lista)
print(f"Suma de la lista: {suma_total}")  # Output: 100

# 4. Elimina el segundo elemento de la lista (índice 1, que es el número 20)
lista.pop(1)

# 5. Añade otro elemento en la posición 3 de la lista (índice 2)
lista.insert(2, 99)

# 6. Crea otra lista con 4 elementos y concaténala a la que ya tenías
otra_lista = [50, 60, 70, 80]
lista.extend(otra_lista)

# --- Estado final de la lista ---
print(f"Lista final resultante: {lista}")
````
