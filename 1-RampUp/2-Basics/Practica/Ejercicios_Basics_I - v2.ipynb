### 🧠 Hipótesis de Sintaxis en Python

A continuación se analiza cada línea según las reglas de nombrado de variables y manejo de tipos de datos en Python:

*   **1. `mi_variable = "Economía"`**  
    * **Estado:** **Correcta**  
    * **Motivo:** Sigue la convención *snake_case* (letras y guiones bajos). El uso de caracteres con tilde en el valor del texto es perfectamente válido.
*   **2. `otra_var = "Ejercicio`**  
    * **Estado:** ❌ **Incorrecta**  
    * **Motivo:** Falta la comilla de cierre (`"`) al final de la cadena de texto, lo que provocará un error de tipo `SyntaxError` (literal de cadena no terminado).
*   **3. `True = "Ejercicio"`**  
    * **Estado:** ❌ **Incorrecta**  
    * **Motivo:** `True` es una palabra clave reservada en Python (representa el valor booleano verdadero). El lenguaje prohíbe reasignar estas palabras como variables.
*   **4. `mi variab1e = "Alpha"`**  
    * **Estado:** ❌ **Incorrecta**  
    * **Motivo:** El nombre de la variable contiene un espacio en blanco. Python no permite espacios para separar el nombre de un mismo identificador.
*   **5. `import = 40`**  
    * **Estado:** ❌ **Incorrecta**  
    * **Motivo:** Al igual que en el caso de `True`, `import` es una palabra clave reservada del sistema (utilizada para cargar librerías) y no se puede usar como variable.
*   **6. `81mi_variable = "Agua"`**  
    * **Estado:** ❌ **Incorrecta**  
    * **Motivo:** Los nombres de las variables en Python pueden contener números, pero **nunca** pueden comenzar con uno.
*   **7. `mi_variable10 = 6`**  
    * **Estado:** **Correcta**  
    * **Motivo:** Cumple todas las reglas: combina letras, guion bajo y finaliza con números, lo cual es totalmente válido.


```python
print("--- COMPROBACIÓN DE VARIABLES ---")

# 1. mi_variable = "Economía"
try:
    mi_variable = "Economía"
    print("✅ Línea 1: ¡Correcta! Guardado:", mi_variable)
except Exception as e:
    print(f"❌ Línea 1: Error -> {type(e).__name__}: {e}")

# 2. otra_var = "Ejercicio
try:
    # Usamos exec porque un error de sintaxis crudo detendría todo el cuaderno
    exec('otra_var = "Ejercicio')
except Exception as e:
    print(f"❌ Línea 2: Error -> {type(e).__name__}: {e}")

# 3. True = "Ejercicio"
try:
    exec('True = "Ejercicio"')
except Exception as e:
    print(f"❌ Línea 3: Error -> {type(e).__name__}: {e}")

# 4. mi variab1e = "Alpha"
try:
    exec('mi variab1e = "Alpha"')
except Exception as e:
    print(f"❌ Línea 4: Error -> {type(e).__name__}: {e}")

# 5. import = 40
try:
    exec('import = 40')
except Exception as e:
    print(f"❌ Línea 5: Error -> {type(e).__name__}: {e}")

# 6. 81mi_variable = "Agua"
try:
    exec('81mi_variable = "Agua"')
except Exception as e:
    print(f"❌ Línea 6: Error -> {type(e).__name__}: {e}")

# 7. mi_variable10 = 6
try:
    mi_variable10 = 6
    print("✅ Línea 7: ¡Correcta! Guardado:", mi_variable10)
except Exception as e:
    print(f"❌ Línea 7: Error -> {type(e).__name__}: {e}")
```

## Ejercicio 4
Realiza un conversor de grados Celsius a Fahrenheit. El conversor tiene que recibir el input del usuario y devolver la conversión.

La conversión de grados sigue la siguiente fórmula: F = C × (9 / 5) + 32 

```python
# 1. Solicitamos los grados Celsius al usuario y los convertimos a número decimal (float)
celsius = float(input("Introduce la temperatura en grados Celsius: "))

# 2. Aplicamos la fórmula matemática para la conversión
fahrenheit = celsius * (9 / 5) + 32

# 3. Mostramos el resultado final por pantalla redondeado a 2 decimales
print(f"\n🌡️ Resultado de la conversión:")
print(f"{celsius}°C equivalen a {fahrenheit:.2f}°F")
```

## Ejercicio 5
Comenta  el Ejercicio 4. Pon un comentario multilinea y haz comentarios de cada linea de código

```python
"""
EJERCICIO 4: Conversor de unidades de temperatura.
Este script solicita al usuario un valor numérico en grados Celsius,
aplica la fórmula de conversión matemática estándar y muestra el
resultado equivalente en la escala Fahrenheit por pantalla.
"""

# Solicitamos la temperatura por teclado al usuario usando la función input()
entrada_usuario = input("Introduce la temperatura en grados Celsius: ")

# Convertimos el texto introducido a un número decimal flotante para poder operar con él
celsius = float(entrada_usuario)

# Aplicamos la fórmula F = C * (9/5) + 32 guardando el resultado en una nueva variable
fahrenheit = celsius * (9 / 5) + 32

# Imprimimos en la terminal una línea de texto decorativa para separar los resultados
print("\n🌡️ Resultado de la conversión:")

# Mostramos la equivalencia final formateando el resultado a un máximo de dos decimales
print(f"{celsius}°C equivalen a {fahrenheit:.2f}°F")
```
