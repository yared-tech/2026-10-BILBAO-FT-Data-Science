## Ejercicio 3
Crea una lista nueva con todas las combinaciones de las siguientes dos listas:
> ```ejer_3_1 = ["Hola", "amigo"]```
>
> ```ejer_3_2 = ["Que", "tal"]```

Obten el siguiente output:

```['Hola Que', 'Hola tal', 'amigo Que', 'amigo tal']```

````python
ejer_3_1 = ["Hola", "amigo"]
ejer_3_2 = ["Que", "tal"]
resultado = [] 

for palabra1 in ejer_3_1: # 2. Primer bucle (externo)
    for palabra2 in ejer_3_2: # 3. Segundo bucle (interno)
        # 4. Junta las palabras con un espacio y las añade a la lista
        resultado.append(f"{palabra1} {palabra2}")

print(resultado)
````

````python
ejer_3_1 = ["Hola", "amigo"]
ejer_3_2 = ["Que", "tal"]

resultado = [f"{palabra1} {palabra2}" for palabra1 in ejer_3_1 for palabra2 in ejer_3_2]

print(resultado)
````

## Ejercicio 11
Convierte la siguiente lista en un diccionario
> ```ejer_11 = [("x", 1), ("x", 2), ("x", 3), ("y", 1), ("y", 2), ("z", 1)]```

TIP: Puedes usar dos variables de control en un bucle `for`. Una por cada valor de las tuplas.

````python
ejer_11 = [("x", 1), ("x", 2), ("x", 3), ("y", 1), ("y", 2), ("z", 1)]

diccionario = {}

# Usamos dos variables de control: 'clave' para la letra y 'valor' para el número
for clave, valor in ejer_11:
    # Si la letra aún no está en el diccionario, le creamos una lista vacía
    if clave not in diccionario:
        diccionario[clave] = []
    
    # Añadimos el número a la lista de esa letra
    diccionario[clave].append(valor)

print(diccionario)
# Output: {'x':, 'y':, 'z':}
````

## Ejercicio 12
1. Crea una lista ordenada ascendente con las claves del diccionario
2. Crea otra lista ordenada descendente con los valores
3. Añade una nueva clave/valor
4. Busca la clave igual a 2 dentro del diccionario
5. Itera la clave y el valor del diccionario con un único for
> ```ejer_12 = {4:78, 2:98, 8:234, 5:29}```

````python
ejer_12 = {4:78, 2:98, 8:234, 5:29}

claves_ordenadas = sorted(ejer_12)
print(f"1. Claves ascendentes: {claves_ordenadas}")

valores_descendientes = sorted(ejer_12.values(), reverse=True)
print(f"2. Valores descendentes: {valores_descendientes}")

ejer_12[10] = 500
print(f"3. Diccionario con nuevo elemento: {ejer_12}")

valor_de_2 = ejer_12[2]
print(f"4. El valor de la clave 2 es: {valor_de_2}")

print("5. Iteración del diccionario:")
for clave, valor in ejer_12.items():
    print(f" Clave: {clave} -> Valor: {valor}")
````

## Ejercicio 16
1. Crea un set de tres elementos
2. Añade un cuarto
3. Elimina el último elemento añadido
4. Elimina el elemento 10, si está presente. Usa `discard()`

````python
mi_set = {5, 8, 12}
print(f"1. Set inicial: {mi_set}")

cuarto_elemento = 25
mi_set.add(cuarto_elemento)
print(f"2. Tras añadir el cuarto: {mi_set}")

mi_set.remove(cuarto_elemento)
print(f"3. Tras eliminar el último añadido: {mi_set}")

mi_set.discard(10)
print(f"4. Tras aplicar discard(10): {mi_set}")
````
