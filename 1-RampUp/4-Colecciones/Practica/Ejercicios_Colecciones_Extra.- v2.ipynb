```python
biblioteca = [
    {"titulo": "Girl, Interrrupted", "autor": "Susanna Keysen", "alquilado": False},
    {"titulo": "A Song Of Ice and Fire", "autor": "George R.R, Martin", "alquilado": False}
]

while True:
    print("\n--- SOFTWARE DE LA BIBLIOTECA ---")
    print("1. Visualizar todos los libros")
    print("2. Buscar libro")
    print("3. Añadir libro")
    print("4. Eliminar libro")
    print("5. Alquilar libro")
    print("6. Devolver libro")
    print("7. Salir")
    
    opcion = input("Elige una opción: ")
    
    if opcion == "7":
        print("Saliendo del programa...")
        break
        
    match opcion:
        case "1":
            if not biblioteca:
                print("La biblioteca está vacía.")
            else:
                for libro in biblioteca:
                    print(f"Título: {libro['titulo']} | Autor: {libro['autor']} | Alquilado: {libro['alquilado']}")
                    
        case "2":
            busqueda = input("Introduce el título del libro a buscar: ")
            encontrado = False
            for libro in biblioteca:
                if libro["titulo"].lower() == busqueda.lower():
                    print(f"Libro encontrado -> Título: {libro['titulo']} | Autor: {libro['autor']} | Alquilado: {libro['alquilado']}")
                    encontrado = True
                    break
            if not encontrado:
                print("El libro no se encuentra en la biblioteca.")
                
        case "3":
            titulo = input("Introduce el título del nuevo libro: ")
            autor = input("Introduce el autor del nuevo libro: ")
            nuevo_libro = {"titulo": titulo, "autor": autor, "alquilado": False}
            biblioteca.append(nuevo_libro)
            print(f"Libro '{titulo}' añadido correctamente.")
            
        case "4":
            titulo_eliminar = input("Introduce el título del libro a eliminar: ")
            encontrado = False
            for libro in biblioteca:
                if libro["titulo"].lower() == titulo_eliminar.lower():
                    biblioteca.remove(libro)
                    print(f"Libro '{libro['titulo']}' eliminado correctamente.")
                    encontrado = True
                    break
            if not encontrado:
                print("No se encontró ningún libro con ese título.")
                
        case "5":
            titulo_alquilar = input("Introduce el título del libro a alquilar: ")
            encontrado = False
            for libro in biblioteca:
                if libro["titulo"].lower() == titulo_alquilar.lower():
                    libro["alquilado"] = True
                    print(f"El libro '{libro['titulo']}' ahora está alquilado.")
                    encontrado = True
                    break
            if not encontrado:
                print("No se encontró ningún libro con ese título.")
                
        case "6":
            titulo_devolver = input("Introduce el título del libro a devolver: ")
            encontrado = False
            for libro in biblioteca:
                if libro["titulo"].lower() == titulo_devolver.lower():
                    libro["alquilado"] = False
                    print(f"El libro '{libro['titulo']}' ha sido devuelto.")
                    encontrado = True
                    break
            if not encontrado:
                print("No se encontró ningún libro con ese título.")
                
        case _:
            print("Opción no válida. Introduce un número del 1 al 7.")
```
