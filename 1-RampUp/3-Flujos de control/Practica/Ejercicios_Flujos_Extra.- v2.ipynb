# Ejercicios Flujos de control

## Ejercicio Extra
Programa una calculadora que

* Pregunte qué operación quieres realizar (multiplicación/división/suma/resta/potencia/salir)
* Pida dos números 
* Realice la operación
* Siga funcionando hasta que el usuario elija salir

#### Bonus:
* Que guarde todas las operaciones realizadas y las imprima por pantalla cuando el usuario elija esa opción
* Que sea robusta frente a entradas erróneas del usuario o frente a operaciones como dividir por 0

```python
import math
import pprint
import time

def obtener_numeros(requiere_dos=True):
    try:
        num1 = float(input("Introduce el primer número (o la base): "))
        num2 = float(input("Introduce el segundo número (o el exponente): ")) if requiere_dos else None
        return num1, num2
    except ValueError:
        print("Error: Datos no válidos.")
        return None

def calcular(operacion, num1, num2=None):
    match operacion:
        case "suma":
            return num1 + num2
        case "resta":
            return num1 - num2
        case "multiplicacion":
            return num1 * num2
        case "division":
            if num2 == 0:
                print("Error: División por cero.")
                return None
            return num1 / num2
        case "potencia":
            try:
                return math.pow(num1, num2)
            except (OverflowError, ValueError):
                print("Error: Operación no válida o resultado demasiado grande.")
                return None
        case "raiz":
            if num1 < 0:
                print("Error: No existe raíz real de un número negativo.")
                return None
            return math.sqrt(num1)

historial = []

while True:

    print("\n--- CALCULADORA AVANZADA ---")
    print("Opciones: suma, resta, multiplicacion, division, potencia, raiz, historial, salir")
    opcion = input("Elige una opción: ").strip().lower()

    match opcion:
        case "salir":
            break

        case "historial":
            if not historial:
                print("El historial está vacío.")
            else:
                print("\n--- HISTORIAL DETALLADO ---")
                pprint.pprint(historial)

        case "suma" | "resta" | "multiplicacion" | "division" | "potencia" | "raiz":
            valores = obtener_numeros(requiere_dos=(opcion != "raiz"))
            if valores is None:
                continue
                
            num1, num2 = valores
            
            print("Calculando...")
            time.sleep(0.5)
            
            resultado = calcular(opcion, num1, num2)
            
            if resultado is not None:
                print(f"Resultado: {resultado}")
                historial.append({
                    "fecha_hora": time.strftime("%Y-%m-%d %H:%M:%S"),
                    "operacion": opcion,
                    "numeros_usados": [num1] if opcion == "raiz" else [num1, num2],
                    "resultado": resultado
                })

        case _:
            print("Opción no válida. Inténtalo de nuevo.")
```
