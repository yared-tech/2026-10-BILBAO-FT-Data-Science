# Extrae los dominios de los correos con una expresión regular

```python
import re

emails = """
miguellc@thebridgeschool.tech
miguel.cruz@uc3m.es
miguel-12-2000@mi-correo.net
miguellc@s@thebridgeschool.es
"""

# Explicación de la Regex definitiva:
# .*@     -> Consume de forma codiciosa (greedy) todo el texto hasta llegar a la ÚLTIMA arroba de la línea
# (       -> Abre el grupo de captura para aislar el dominio real
#  [a-zA-Z0-9.-]+ -> Captura caracteres válidos de dominio (letras, números, guiones y puntos)
# )       -> Cierra el grupo de captura
patron_avanzado = r".*@([a-zA-Z0-9.-]+)"

# Al no usar los anclajes rígidos de inicio de línea (^), procesa el final de forma limpia
dominios = re.findall(patron_avanzado, emails)

print(dominios)
```
