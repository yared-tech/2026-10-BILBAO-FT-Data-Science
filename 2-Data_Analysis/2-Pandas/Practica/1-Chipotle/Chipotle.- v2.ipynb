### Step 1. Import the necessary libraries

```python
import pandas as pd
import numpy as np
```

### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv) and assign it to a variable called chipo.

```python
url = "https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv"
chipo = pd.read_csv(url, sep="\t")
chipo
```

```python
# 1. Limpiar la columna quitando el '$' y convirtiéndola a número flotante
chipo["item_price"] = chipo["item_price"].str.replace("$", "").astype(float)

# 2. Encontrar el índice de la fila con el precio máximo y sacar su 'item_name'
max_product_name = chipo.loc[chipo["item_price"].idxmax(), "item_name"]

print(max_product_name)
```

### Step 4. How many products cost more than $10.00?

```python
chipo['item_price'] = chipo['item_price'].astype(str).str.replace('$', '', regex=False).astype(float)

resultado = chipo[chipo['item_price'] > 10.0]
print(resultado.head())
```

### Step 4.1: Y cuántos pedidos se han hecho con un producto de más de 10$? Es lo mismo?

```python

chipo['item_price'] = chipo['item_price'].astype(str).str.replace('$', '', regex=False).astype(float)

filtro_caros = chipo[(chipo['item_price'] / chipo['quantity']) > 10.0]

print(filtro_caros)

total_pedidos = filtro_caros['order_id'].nunique()

print(total_pedidos)

print(f"Se han hecho {total_pedidos} pedidos que contienen al menos un producto de más de $10.")
```

### Step 4.2: Y cuántos pedidos se han hecho de más de 10$? Es lo mismo?

```python

chipo['item_price'] = chipo['item_price'].astype(str).str.replace('$', '', regex=False).astype(float)

ticket_por_pedido = chipo.groupby('order_id')['item_price'].sum()

pedidos_mas_de_10 = ticket_por_pedido[ticket_por_pedido > 10.0].count()

print(f"Se han hecho {pedidos_mas_de_10} pedidos cuyo ticket total superó los $10.")
```

### Step 4.3: Y en cuántos pedidos se ha pagado más de 10$ por un mismo producto? Es lo mismo?

```python

chipo['item_price'] = chipo['item_price'].astype(str).str.replace('$', '', regex=False).astype(float)

filtro_renglon_caro = chipo[chipo['item_price'] > 10.0]

pedidos_mismo_producto_mas_10 = filtro_renglon_caro['order_id'].nunique()

print(f"Se han hecho {pedidos_mismo_producto_mas_10} pedidos donde se pagó más de $10 por un mismo producto.")
```

### Step 5. What is the price of each item and name it unit_price. Get only item_name and unit_price

```python
if chipo['item_price'].dtype == 'object':
    chipo['item_price'] = chipo['item_price'].str.replace('$', '', regex=False).astype(float)

chipo['unit_price'] = chipo['item_price'] / chipo['quantity']

productos_unicos = chipo.drop_duplicates(['item_name', 'choice_description'])

resultado_final = productos_unicos[['item_name', 'unit_price']]

print(resultado_final.head(10))
```

### Step 6. Sort by the name of the item

```python
if chipo['item_price'].dtype == 'object':
    chipo['item_price'] = chipo['item_price'].str.replace('$', '', regex=False).astype(float)

chipo['unit_price'] = chipo['item_price'] / chipo['quantity']

productos_unicos = chipo.drop_duplicates(['item_name', 'choice_description'])

resultado_final = productos_unicos[['item_name', 'unit_price']]

resultado_ordenado = resultado_final.sort_values(by='item_name')

print(resultado_ordenado.head(15))
```

### Step 7. What was the quantity of the most expensive item ordered? 2 ways

```python
# ==========================================
# PASO PREVIO OBLIGATORIO: LIMPIEZA DE DATOS
# ==========================================
# Aseguramos que la columna sea numérica eliminando el '$' si es texto
if chipo['item_price'].dtype == 'object':
    chipo['item_price'] = chipo['item_price'].astype(str).str.replace('$', '', regex=False).astype(float)

# Calculamos el precio unitario (el precio de una sola unidad en la carta)
chipo['unit_price'] = chipo['item_price'] / chipo['quantity']

print(chipo['unit_price'])

# ==========================================
# MANERA 1: Por el mayor precio UNITARIO
# (El plato más caro de la carta)
# ==========================================
# Usamos .argmax() con .iloc para buscar por posición física y evitar KeyErrors
posicion_max_unit = chipo['unit_price'].argmax()
producto_max_unit = chipo.iloc[posicion_max_unit]['item_name']
precio_max_unit = chipo.iloc[posicion_max_unit]['unit_price']
cantidad_manera1 = chipo.iloc[posicion_max_unit]['quantity']

print("==========================================================")
print("MANERA 1: BASADO EN EL PLATO MÁS CARO DEL MENÚ")
print("==========================================================")
print(f"• Producto más caro  : {producto_max_unit}")
print(f"• Precio unitario    : ${precio_max_unit:.2f}")
print(f"• Cantidad ordenada  : {cantidad_manera1}")
print("==========================================================\n")


# ==========================================
# MANERA 2: Por el mayor valor de renglón
# (La compra/gasto más grande de golpe)
# ==========================================
# Buscamos la posición del 'item_price' total más alto registrado
posicion_max_total = chipo['item_price'].argmax()
producto_max_total = chipo.iloc[posicion_max_total]['item_name']
gasto_max_total = chipo.iloc[posicion_max_total]['item_price']
cantidad_manera2 = chipo.iloc[posicion_max_total]['quantity']

print("==========================================================")
print("MANERA 2: BASADO EN LA TRANSACCIÓN MÁS GRANDE REGISTRADA")
print("==========================================================")
print(f"• Producto implicado : {producto_max_total}")
print(f"• Gasto total fila   : ${gasto_max_total:.2f}")
print(f"• Cantidad ordenada  : {cantidad_manera2}")
print("==========================================================")
```

### Step 8. How many times was a Veggie Salad Bowl ordered?

```python
veggie_salad_df = chipo[chipo['item_name'] == "Veggie Salad Bowl"]

veces_ordenado = len(veggie_salad_df)

total_unidades = veggie_salad_df['quantity'].sum()

print(f"• Aparece en {veces_ordenado} renglones de pedidos.")
print(f"• Se ordenaron {total_unidades} unidades en total.")
```
