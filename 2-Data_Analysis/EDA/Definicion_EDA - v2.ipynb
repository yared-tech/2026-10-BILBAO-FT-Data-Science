# INFORMACIÓN DEL PROYECTO

### TÍTULO

Radiografía genómica de la Era Vikinga: Contraste empírico entre el mito de la homogeneidad nórdica, las rutas de mestizaje y los patrones de movilidad por género

### TEMA

Estudio exploratorio de datos (EDA) basado en el análisis paleogenómico de 442 individuos antiguos y más de 1.500 genomas de referencia (Margaryan et al., Nature 2020) para evaluar la estructura poblacional real de la sociedad vikinga. A través del cruce de metadatos geográficos, proporciones de ascendencia por bootstrapping, modelos de mezcla demográfica y linajes uniparentales por género, el proyecto investiga cómo el fenómeno vikingo no constituyó una población aislada ni homogénea, sino una red transregional marcada por un intenso flujo génico desde el sur y este de Europa, dinámicas territoriales diferenciadas según la ruta (Atlántico vs. Báltico) y patrones de movilidad demográfica heterogéneos entre hombres y mujeres.

### HIPÓTESIS

**Hipótesis principal** La sociedad de la Era Vikinga no constituyó un grupo genéticamente aislado ni homogéneo: más del 56% de los individuos analizados en yacimientos de la propia Escandinavia presentan una ascendencia genética foránea superior al 30% (procedente principalmente del sur y este de Europa y de las Islas Británicas). Esto demuestra que el fenómeno vikingo operaba como una red comercial y migratoria altamente conectada, refutando el mito de la pureza nórdica.

**Hipótesis Secundaria 1** (Direccionalidad geográfica - Oeste vs. Este):
Los patrones de mestizaje varían drásticamente según la ruta de expansión geográfica; los asentamientos del Atlántico Norte (como Islandia e Irlanda) muestran una fuerte correlación con linajes británicos y noratlánticos (superando el 41% de media), mientras que los enclaves orientales (como Polonia o Ucrania) absorben de manera prioritaria sustratos eslavos y fino-bálticos (por encima del 54%).

**Hipótesis Secundaria 2** (El sur escandinavo como crisol comercial):
Las regiones costeras y portuarias del sur de Escandinavia (Dinamarca y sur de Suecia) concentran los niveles más altos de diversidad genética y una mayor proporción de individuos con aportes meridionales (sur de Europa y Mediterráneo), reflejando su función como centros neurálgicos de intercambio económico y cultural frente a las zonas del interior.

**Hipótesis Secundaria 3** (Movilidad y mestizaje por género):
Existen diferencias significativas en la dinámica poblacional según el sexo; los enterramientos en contextos de incursiones y expediciones militares muestran una abrumadora predominancia masculina y alta homogeneidad en linajes paternos, mientras que en los asentamientos coloniales estables del Atlántico Norte se observa una gran diversidad de linajes maternos foráneos, reflejando una integración demográfica mediada por la incorporación de mujeres locales.
